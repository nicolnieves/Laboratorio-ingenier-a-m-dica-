```python

import pandas as pd

URL = "https://raw.githubusercontent.com/nicolnieves/Laboratorio-ingenier-a-m-dica-/6fdbbede106c6049752312e55236821a058304d2/pacientes_simulados_200_53843a0f20be084cc2104e81b2d30bf1.csv"

if not URL.startswith("https://raw.githubusercontent.com/"):
    raise ValueError("Pega la URL RAW del CSV")

df = pd.read_csv(URL)

print("Archivo conectado. Filas y columnas:", df.shape)

display(df.head())
```

```python
print("Filas:", len(df))
print("Columnas:", len(df.columns))
print("Nombres de columnas:", list(df.columns))

display(df.sample(5, random_state=7))
```

```python
print("Datos faltantes por columna:")
print(df.isna().sum())

print("Identificadores duplicados:")
print(df["Paciente"].duplicated().sum())

print("Tipos de dato:")
print(df.dtypes)
```

```python
columnas = [
    "Edad",
    "FrecuenciaCardiaca",
    "Saturacion",
    "Temperatura",
    "Glucosa"
]

display(df[columnas].describe().round(2))

print("Edad media:", round(df["Edad"].mean(), 1))
print("Edad mínima:", df["Edad"].min())
print("Edad máxima:", df["Edad"].max())
```

```python
criterio_fc = 95

seleccion = df[df["FrecuenciaCardiaca"] > criterio_fc]

print("Registros que superan el valor de prueba:", len(seleccion))

display(seleccion[
    ["Paciente", "Edad", "FrecuenciaCardiaca"]
].head(15))
```

```python
criterio_sat = 96

revision = df[
    (df["Saturacion"] < criterio_sat) &
    (df["FrecuenciaCardiaca"] > 100)
]

print("Registros que cumplen ambas condiciones:", len(revision))

display(revision[
    ["Paciente", "FrecuenciaCardiaca", "Saturacion"]
])
```

```python
df["IMC"] = (
    df["Peso"] / df["Estatura"]**2
).round(2)

display(df[[
    "Paciente",
    "Peso",
    "Estatura",
    "IMC"
]].head(12))

print("IMC promedio:", round(df["IMC"].mean(), 2))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(11,4))

plt.plot(df["Paciente"].head(30), df["Saturacion"].head(30), marker="o")

plt.xticks(rotation=90)
plt.title("saturación — primeros 30 registros")
plt.xlabel("Paciente")
plt.ylabel("saturación (%)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.hist(df["Edad"], bins=12, edgecolor="black")
plt.title("Distribución de edades sintéticas")
plt.xlabel("Edad (años)")
plt.ylabel("Cantidad de registros")
plt.show()
```

```python
print("Registros por servicio:")
display(df["Servicio"].value_counts().to_frame("Cantidad"))

resumen = df.groupby("Servicio")[["Edad","FrecuenciaCardiaca","Saturacion"]].mean().round(1)

display(resumen)

resumen["FrecuenciaCardiaca"].plot(
    kind="bar",
    title="Media de frecuencia cardíaca por servicio"
)

plt.ylabel("Latidos por minuto")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```
